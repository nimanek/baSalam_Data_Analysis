# Question from COO
### When is the top purchase time that we can give better support?

```python
import pandas as pd

# Getting products data only 5000 rows
df_products = pd.read_csv('../data/BaSalam.products.csv', nrows=5000)
pd.set_option("display.max_columns", None)

df_products
```

```python
prep_counts = df_products['preparationDays'].value_counts().sort_index()
prep_counts
```

```python
import matplotlib.pyplot as plt

# creating chart
plt.figure(figsize=(8, 4))
plt.bar(prep_counts.index, prep_counts.values, color='skyblue', edgecolor='black')

plt.xlabel('Preparation Days')
plt.ylabel('Number of Products')
plt.title('Distribution of Product Preparation Days')

plt.grid(axis='y', ls='--', alpha=0.5)

# save the chart pic to the outpurs folder and print it here
plt.savefig('../outputs/preparation_days_distribution.png', bbox_inches='tight')
plt.show()
```

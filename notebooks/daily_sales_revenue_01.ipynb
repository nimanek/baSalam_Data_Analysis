# Question from CEO
### How much is our daily revenue?

```python
import pandas as pd

# Getting products data only 5000 rows
df_products = pd.read_csv('../data/BaSalam.products.csv', nrows=5000)

df_products.head()
```

##  Converting Currency =>

```python
# converting price currency from IRR to USD (1 USD = 1.500.000 IRR = 150.000 TOMAN )
df_products['price'] = ((df_products['price'] / 10) / 150000).round(2)
df_products['price']
```

## Daily Revenue

```python
df_products['daily_sale'] = ((df_products['price'] * df_products['sales_count_week']) / 7).sum().round(2)
```

```python
df_products['daily_sale']
```

## Answer to CEO:
### Estimated daily revenue is 872,89$
